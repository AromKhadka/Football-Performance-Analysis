```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("../data/raw/database.sqlite")
```

```python
tables = pd.read_sql_query("""
SELECT name
FROM sqlite_master
WHERE type='table'
ORDER BY name;
""", conn)

tables
```

```python
table_names = tables["name"].tolist()

for table in table_names:
    print(f"\n--- {table.upper()} ---")
    df = pd.read_sql_query(f'SELECT * FROM "{table}" LIMIT 5;', conn)
    print(df.head())
    print(df.columns.tolist())
    print("Shape preview not shown here; use count query below.")
```

```python
for table in table_names:
    count = pd.read_sql_query(f'SELECT COUNT(*) AS row_count FROM "{table}";', conn)
    print(f"{table}: {count.loc[0, 'row_count']}")
```

```python
important_tables = ["country", "league", "team", "player", "player_attributes", "match"]

for table in important_tables:
    print(f"\n--- {table.upper()} SCHEMA ---")
    schema = pd.read_sql_query(f'PRAGMA table_info("{table}");', conn)
    print(schema)
```

# Data Overview Notes

This project uses a SQLite database containing football data across several tables.  
The main tables for this analysis are:

- **match**: match results and game-level statistics
- **team**: team names and identifiers
- **league**: league names and country mapping
- **country**: country names
- **player** and **player_attributes**: optional tables for player-level analysis later

For the first version of the project, the focus will be on **match, team, league, and country** data.
