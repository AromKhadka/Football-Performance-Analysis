```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("../data/cleaned/football_analysis.db")
```

```python
df = pd.read_csv("../data/cleaned/football_matches_cleaned.csv")
df.head()
```

```python
df.to_sql("matches", conn, if_exists="replace", index=False)
```

```python
pd.read_sql("SELECT * FROM matches LIMIT 5;", conn)
```

```python
pd.read_sql("""
SELECT COUNT(*) AS total_matches
FROM matches;
""", conn)
```

```python
pd.read_sql("""
SELECT ROUND(AVG(total_goals), 2) AS avg_goals
FROM matches;
""", conn)
```

```python
pd.read_sql("""
SELECT result, COUNT(*) AS count
FROM matches
GROUP BY result;
""", conn)
```

```python
pd.read_sql("""
SELECT 
    ROUND(100.0 * SUM(CASE WHEN result = 'Home Win' THEN 1 ELSE 0 END) / COUNT(*), 2) AS home_win_percentage
FROM matches;
""", conn)
```

```python
pd.read_sql("""
SELECT 
    league_name,
    ROUND(AVG(total_goals), 2) AS avg_goals
FROM matches
GROUP BY league_name
ORDER BY avg_goals DESC;
""", conn)
```

```python
pd.read_sql("""
SELECT 
    season,
    COUNT(*) AS total_matches
FROM matches
GROUP BY season
ORDER BY season;
""", conn)
```

```python
pd.read_sql("""
SELECT 
    home_team_name,
    away_team_name,
    total_goals
FROM matches
ORDER BY total_goals DESC
LIMIT 10;
""", conn)
```

```python
pd.read_sql("""
SELECT 
    home_team_name AS team,
    COUNT(*) AS wins
FROM matches
WHERE result = 'Home Win'
GROUP BY home_team_name
ORDER BY wins DESC
LIMIT 10;
""", conn)
```

```python
pd.read_sql("""
SELECT 
    away_team_name AS team,
    COUNT(*) AS wins
FROM matches
WHERE result = 'Away Win'
GROUP BY away_team_name
ORDER BY wins DESC
LIMIT 10;
""", conn)
```

```python
pd.read_sql("""
SELECT 
    home_team_name AS team,
    SUM(home_team_goal) AS goals_scored
FROM matches
GROUP BY home_team_name
ORDER BY goals_scored DESC
LIMIT 10;
""", conn)
```

```python
pd.read_sql("""
SELECT 
    home_team_name AS team,
    SUM(away_team_goal) AS goals_conceded
FROM matches
GROUP BY home_team_name
ORDER BY goals_conceded ASC
LIMIT 10;
""", conn)
```

```python
pd.read_sql("""
SELECT 
    home_team_name AS team,
    COUNT(*) AS matches,
    SUM(CASE WHEN result = 'Home Win' THEN 1 ELSE 0 END) AS wins,
    ROUND(
        100.0 * SUM(CASE WHEN result = 'Home Win' THEN 1 ELSE 0 END) / COUNT(*), 2
    ) AS win_rate
FROM matches
GROUP BY home_team_name
HAVING matches > 50
ORDER BY win_rate DESC
LIMIT 10;
""", conn)
```

```python
pd.read_sql("""
SELECT 
    season,
    ROUND(AVG(total_goals), 2) AS avg_goals
FROM matches
GROUP BY season
ORDER BY season;
""", conn)
```

```python
pd.read_sql("""
SELECT 
    ROUND(
        100.0 * SUM(CASE WHEN result = 'Draw' THEN 1 ELSE 0 END) / COUNT(*), 2
    ) AS draw_percentage
FROM matches;
""", conn)
```

```python
pd.read_sql("""
SELECT 
    league_name,
    ROUND(
        100.0 * SUM(CASE WHEN result = 'Draw' THEN 1 ELSE 0 END) / COUNT(*), 2
    ) AS draw_rate
FROM matches
GROUP BY league_name
ORDER BY draw_rate DESC;
""", conn)
```

# SQL Analysis Summary

Key findings from SQL analysis:

- Home teams win a higher percentage of matches, confirming home advantage
- Some leagues consistently produce more goals than others
- Top-performing teams dominate both home and away matches
- Certain teams show strong defensive performance with low goals conceded
- Match outcomes and scoring patterns vary across seasons

These insights support deeper visualization in Power BI and can guide predictive modeling.
